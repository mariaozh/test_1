## 4.Cамое популярное мужское и самое популярное женское имя на корабле

```python
# Как прописано мужское имя
df.loc[df['Sex']=='male','Name']
```

```python
# Как прописано женское имя
df.loc[df['Sex']=='female','Name']
```

```python
# Функция для выдедения мужского имени из строки
def parse_first_name_male(name):
    first = name.str.extract(r"Mr\.\s+[^(]*\((\w+)", expand=False)
    first.loc[first.isna()] = name.str.extract(r"\.\s+(\w+)", expand=False)
    return first

# Функция для выдедения женского имени
def parse_first_name_female(name):
    first = name.str.extract(r"Mrs\.\s+[^(]*\((\w+)", expand=False)
    first.loc[first.isna()] = name.str.extract(r"\.\s+(\w+)", expand=False)
    return first
```

```python
# Выделенные мужских имен
male_names = parse_first_name_male(df.loc[df['Sex']=='male','Name'])
male_names
```

```python
# Популярность мужского имени
male_names.value_counts()
```

```python
# Выделенные женские имена
female_names = parse_first_name_female(df.loc[df['Sex']=='female','Name'])
female_names
```

```python
# Популярность женского имени
female_names.value_counts()
```

## 5.Cамое популярное мужское и самое популярное женское имя на корабле в каждом классе

```python
#Часть базы с пассажирами мужского пола
df_male=df[df['Sex'] == 'male']
df_male = df_male.rename(columns={'Name': 'Ful_Name'})
```

```python
#Объединение таблицы с данными пасссажиров мужского пола и таблицей выделенных имен
df_male_names = pd.concat([df_male, male_names], sort=False, axis=1)
df_male_names
```

```python
# Популярность мужского имени по классам
df_res5_1 = df_male_names.groupby(["Pclass"])["Name"].value_counts()
df_res5_1
```

```python
#Часть базы с пассажирами женского пола
df_female=df[df['Sex'] == 'female']
df_female = df_female.rename(columns={'Name': 'Ful_Name'})
```

```python
#Объединение таблицы с данными пасссажиров женского пола и таблицей выделенных имен
df_female_names = pd.concat([df_female, female_names], sort=False, axis=1)
df_female_names
```

```python
# Популярность женского имени по классам
df_res5_2 = df_female_names.groupby(["Pclass"])["Name"].value_counts()
df_res5_2
```

## 8.Количества *n* - местных кабин (в которых было 2, 3, 4, ... человека)

```python
# Подсчет элементов с одинаковыми значениями в том или ином столбце
df['Cabin'].value_counts()
```

```python
# Количество человек в каждой кабине согласно базе данных
df_cabin=df.groupby("Cabin")["Cabin"].value_counts()
df_cabin_people = df_cabin.to_frame(name='Capacity')
df_cabin_people
```

```python
#Количества n-местных кабин
df_cabin_people.groupby(["Capacity"])["Capacity"].value_counts()
```
